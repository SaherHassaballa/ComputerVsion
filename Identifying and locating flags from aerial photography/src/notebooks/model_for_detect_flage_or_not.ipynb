```python
from supervision import Detections, BoxAnnotator
from supervision.detection.tools.inference_slicer import InferenceSlicer
import cv2
```

```python
metrics = model.val()
metrics
```

```python
def slicer_callback(image_slice: cv2.Mat) -> Detections:
    res = model(image_slice)[0]
    return Detections.from_ultralytics(res)

slicer = InferenceSlicer(
    callback=slicer_callback,
    slice_wh=(640, 640),             # حجم كل قطعة
    overlap_ratio_wh=(0.2, 0.2),     # تداخل 20%
    iou_threshold=0.5,
    thread_workers=4
)
```

```python
def detect_with_sahi(image_path: str):
    img = cv2.imread(image_path)
    detections = slicer(image=img)
    annotator = BoxAnnotator()
    annotated = annotator.annotate(scene=img.copy(), detections=detections)
    output_path = image_path.replace('.jpg', '_result.jpg')
    cv2.imwrite(output_path, annotated)
    print(f"✅ تم حفظ: {output_path} — عدد الصناديق: {len(detections.xyxy)}")
```
