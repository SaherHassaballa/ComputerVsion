```python
import gc, torch
gc.collect()
torch.cuda.empty_cache()

from ultralytics import YOLO

path_yolo = r"C:\Users\saher\Desktop\github\computer vision\projects\Identifying and locating flags from aerial photography\src\notebooks\yolo11l.pt"

model = YOLO(path_yolo)
```

```python
path_yaml = r"D:\github lite\cv_projects\Identifying and locating flags from aerial photography\data\backgrounds\data.yaml"

model.train(
    data=path_yaml,      # مسار ملف البيانات
    epochs=30,
    batch=1,              # اضبطه حسب VRAM المتاح
    imgsz=1024,            # حجم صورة أكبر لتحسين التفاصيل
    lr0=0.001,
    optimizer='AdamW',
    augment=True,
    hsv_h=0.02, hsv_s=0.5, hsv_v=0.5,
    translate=0.2, scale=0.5,
    flipud=0.5, fliplr=0.5,
    mosaic=1.0, mixup=0.5, copy_paste=0.5,
    multi_scale=True,
    amp=True,
    patience=5,
    device=0,               # استخدم GPU
    project = r"D:\github lite\cv_projects\Identifying and locating flags from aerial photography\models\trained_models",
    name="kosmak"
)
```
