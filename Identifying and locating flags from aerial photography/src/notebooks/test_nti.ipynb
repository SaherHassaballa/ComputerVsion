```python
import os
import cv2 
import numpy as np

images = []
for folder in os.listdir("D:\work space\CV_NTI\dataset"):
    for file in os.listdir(os.path.join( "D:\work space\CV_NTI\dataset",folder)):
        path = os.path.join(os.path.join( "D:\work space\CV_NTI\dataset",folder),file)
        img = cv2.imread(path)
        img = img.astype(np.float32)
        images.append(img)

print(images[0][0])

```

```python
normalization_images = [ img / 255 for img in images]
print(normalization_images[0][0])
```

```python
for img in normalization_images :
    print(img.shape)
```

```python
resizing_images = [ cv2.resize(img , (int(img.shape[1]*0.5) , int(img.shape[0]*0.5))) for img in normalization_images] # take integer for size pixels
for img in resizing_images :
    print(img.shape)
```

لو الداتا قليله اعمل اوجمنتاشن لو في بايس ده هيعمل مشاكل في الموديل هيبقا منحاز فبنعمل بالنس عن طريق الاوزان

```python
resizing_images = [cv2.cvtColor(img , cv2.COLOR_BGR2RGB) for img in resizing_images]
```

```python
import matplotlib.pyplot as plt
fig = plt.figure(figsize=(8, 4))
for i, img in enumerate(resizing_images):
    plt.subplot(2, len(resizing_images)//2, i + 1) # i+1 index of img will display it
    plt.imshow(img)
    plt.axis('off')
plt.show()
```

```python
fig, ax = plt.subplots(2, 4, figsize=(12, 6))
for i, img in enumerate(resizing_images):
    row = i // 4
    col = i % 4
    ax[row, col].imshow(img)
    ax[row, col].axis('off')
plt.tight_layout()
plt.show()
```
